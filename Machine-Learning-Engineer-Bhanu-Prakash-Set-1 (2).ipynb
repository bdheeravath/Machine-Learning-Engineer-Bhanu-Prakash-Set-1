```python
Coding Q1 - Find Peak Element
Problem Statement
A peak element in an array is an element that is strictly greater than its neighbours.
Given a 0-indexed integer array nums, write a function find_peak_element(nums) that returns the index
of any one peak element.
You may assume that nums[-1] = nums[n] = -infinity (elements beyond boundaries are negative infinity).
Your solution must run in O(log n) time. If the array contains multiple peaks, returning any one index is
acceptable.
Constraints
 1 &lt;= nums.length &lt;= 1000
 -2^31 &lt;= nums[i] &lt;= 2^31 - 1
 nums[i] != nums[i+1] for all valid i
Open Test Cases (Visible to Candidate)
Input: nums = [1, 2, 3, 1]
Output: 2 (index 2, value 3 is a peak)
Input: nums = [1, 2, 1, 3, 5, 6, 4]
Output: 5 (index 5, value 6 is a peak; index 1 also valid)
```

```python
Coding Q2 - Group Anagrams
Problem Statement
Given an array of strings strs, group all strings that are anagrams of each other into the same group
and return the list of groups.
Two strings are anagrams if they contain the same characters with the same frequencies.
The output groups and strings within each group may be returned in any order.
Constraints
 1 &lt;= strs.length &lt;= 10^4
 0 &lt;= strs[i].length &lt;= 100
 strs[i] consists of lowercase English letters only
 Expected time complexity: O(n x k log k) where k is the max string length
Open Test Cases (Visible to Candidate)
Input: strs = [&quot;eat&quot;,&quot;tea&quot;,&quot;tan&quot;,&quot;ate&quot;,&quot;nat&quot;,&quot;bat&quot;]
Output: [[&quot;bat&quot;],[&quot;nat&quot;,&quot;tan&quot;],[&quot;ate&quot;,&quot;eat&quot;,&quot;tea&quot;]] (any group order)

Input: strs = [&quot;&quot;]
Output: [[&quot;&quot;]]
```

```python
from collections import defaultdict
 
def group_anagrams(strs):
    groups = defaultdict(list)
 
    for word in strs:
        key = ''.join(sorted(word))
    groups[key].append(word)
 
return list(groups.values())
 
 
# Test Cases
print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
print(group_anagrams([""]))
```
